```python
def conccn(ds):
    ds = ( ds.assign
          (conccn = lambda ds : (((1-lognorm(1.59,ds.RDRY_NUS[:,:]*1e9).cdf(3.0))*ds.N_NUS[:,:]) + ds.N_AIS[:,:] + ds.N_ACS[:,:] + ds.N_COS[:,:] + ds.N_AII[:,:] + ds.N_ACI[:,:] + ds.N_COI[:,:]) * 1e-6))
    return(ds)
```
